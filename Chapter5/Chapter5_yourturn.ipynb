#  Hypothesis Testing

## <font color=#30a64d>Question</font>

In the previous week, we used the datafile [worldpopulation.txt](https://github.com/itrharrison/PX4146/blob/main/Chapter4/worldpopulation.txt) to estimate the correlation between the world population size measured over time.

1. Measure the correlation coefficient using the in built Python package for the Spearman $\rho$ statistic.

2. Measure it again but this time creating the functions to calculate $\rho$ from scratch.   You can use the in-built function to rank datasets from scipy.stats: `from scipy.stats import rankdata`.

### Answer

```python
import numpy as np
f = open("worldpopulation.txt")
worldpop = f.read()

#print(worldpop)
#want to seperate up the string into interger
values = (worldpop.split(' ')) #breakout the string into individual components
values = np.array(np.genfromtxt('worldpopulation.txt')).flatten() #use a numpy method to get strings converted to floats then use flatten to get it in 1D

n = len(values)/2 #need to seperate out the year and population information, find the midpoint to do so
#print(n)
years = values[:int(n)]          #years
pops = values[int(n):]     #population

#built in spearman p
from scipy.stats import pearsonr
from scipy.stats import spearmanr
stats = spearmanr(years,pops)
print(stats)

#now find it from scratch

#rank them to remove the value information, we are just intrested in the pattern
from scipy.stats import rankdata

rank_years = rankdata(years)
rank_pops = rankdata(pops)

sum_xy = np.sum(rank_years*rank_pops)
sum_x = np.sum(rank_years)
sum_y = np.sum(rank_pops)
sum_x_sq = np.sum(rank_years**2)
sum_y_sq = np.sum(rank_pops**2)
n = len(rank_years)

r = (n*sum_xy - sum_x*sum_y) /np.sqrt((n*sum_x_sq - sum_x**2)*(n*sum_y_sq - sum_y**2))

print(r)

print("All the rankings are in order so we get a perfect posotive 1. Every year the population increases.")
```
