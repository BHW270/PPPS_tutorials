## Question

You want to send a document to your bank and it is very important it gets there the next day. You therefore send two copies of the document using Hermes (delivery service $A$) and the Royal Mail (delivery service $B$).  The probabilities of those companies delivering on time are:

- 0.90 for service A
- 0.80 for service B
- 0.75 for both services being on time.

Does your strategy improve the chances the document will be delivered on time?

### Answer

*Your answer here.*

```python
pa = 0.9
print("P(A) = ",  pa)

pb = 0.8
print("P(B) =", pb)

print("Told that P(A) U P(B) = 0.75")
print("Assuming these are independent P(A) U P(B)", (pa*pb), " we get a different value.")#

print("So these are not indpendent")

p_single_method = pa
print("The odds with the singe send method are ", p_single_method )

#to find the combined odds we need to use the addition rule, add up P(A)+P(B) minus the intersection to prevent double counting P(A U B)
#to find the intersection of dependent results or in general, P(A n B) = P(A|B)P(B)

p_a_n_b = 0.75 # we can skip the addition rule as this result is given to us

p_double_method = pa + pb - p_a_n_b

print("odds of at least one arriving on time if sending with both services", p_double_method)

if p_double_method > p_single_method:
  print("The double method does inccrease the chances of the document ariving on time")

else:
  print("The doube method does not increase the chances of the document arriving on time")
```
