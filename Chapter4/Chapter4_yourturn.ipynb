# Errors

## Question

Three students measure the same resistance several times, with final measurements (in Ohms):

Student 1: $R = 11 \pm 1$
Student 2: $R = 12 \pm 1$
Student 3: $R = 10 \pm 3$.

Given these three results what do we write down as the best estimate for the resistance $R$? Tip: you will need to use weighted errors.

### Answer

*Your answer here.*

```python
import numpy as np
R= np.array([11,12,10])
error = np.array([1,1,3])

#w_i = (1/sigma_i)^2
weights = []
for i in error:
  weights.append((1/i)**2)

weights = np.array(weights)

#apply the mean based on the weight
#muitply the measuremnt based on the weight, sum them then divide by the weight
result = (np.sum(weights*R))/np.sum(weights)
print(result)

#unceratinty is the  reciprical sqrt of the sum of the weights
uncertainty = 1/np.sqrt(np.sum(weights))
print(uncertainty)
```
